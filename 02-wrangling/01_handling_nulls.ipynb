# 🧹 Handling Nulls and Missing Values
> **Phase 2 — Data Wrangling** | `data-analysis-practice`

Real-world data is messy. Missing values are the #1 problem you'll face. Learn to detect and handle them properly.

---

```python
import pandas as pd
import numpy as np
```

1. Creating Data with Missing Values

```python
data = {
    "name":   ["Lia", "Juan", "Maria", None, "Ana"],
    "age":    [21, None, 20, 23, 21],
    "score":  [88, 92, None, 95, None],
    "course": ["BSIT", "BSCS", None, "BSCE", "BSCS"]
}
df = pd.DataFrame(data)
print(df)
```

2. Detecting Missing Values

```python
# Check which cells are null
print(df.isnull())
```

```python
# Count nulls per column
print(df.isnull().sum())
```

```python
# Percentage of missing data
print((df.isnull().sum() / len(df)) * 100)
```

3. Dropping Missing Values

```python
# Drop rows with ANY null
df_dropped = df.dropna()
print(df_dropped)
```

```python
# Drop rows only if ALL values are null
df_dropped_all = df.dropna(how="all")
print(df_dropped_all)
```

```python
# Drop rows where specific column is null
df_no_score = df.dropna(subset=["score"])
print(df_no_score)
```

4. Filling Missing Values

```python
# Fill with a specific value
df_filled = df.fillna("Unknown")
print(df_filled)
```

```python
# Fill numeric columns with mean
df["score"] = df["score"].fillna(df["score"].mean())
df["age"] = df["age"].fillna(df["age"].median())
print(df)
```

```python
# Forward fill (use previous row's value)
df_ffill = df.fillna(method="ffill")
print(df_ffill)
```

5. Real-World Example

```python
# Simulate a real messy dataset
import numpy as np
np.random.seed(42)

real_data = pd.DataFrame({
    "student_id": range(1, 11),
    "math":    [85, np.nan, 90, 78, np.nan, 88, 92, np.nan, 76, 85],
    "english": [78, 88, np.nan, 82, 90, np.nan, 85, 79, np.nan, 91],
    "science": [90, 85, 87, np.nan, 88, 92, np.nan, 84, 90, np.nan]
})

print("Missing values:")
print(real_data.isnull().sum())
print()

# Fill with column mean
real_data_clean = real_data.fillna(real_data.mean(numeric_only=True).round(1))
print("After filling with mean:")
print(real_data_clean)
```
