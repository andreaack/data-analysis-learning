```python
import matplotlib.pyplot as plt
import numpy as np

# Example of using plt.show()
x_example = np.array([1, 2, 3])
y_example = np.array([4, 5, 6])
plt.plot(x_example, y_example)
plt.title("Example Plot for plt.show()")
plt.show()
```

```python
#Graph Values

x=np.linspace(0,10,100)
y=np.sin(x)
z=np.cos(y)
```

PLOTTING DATA

```python
print("sine wave")
plt.plot(x,y)
plt.show()
```

```python
print("cosine wave")
plt.plot(x,z)
plt.show()
```

```python
print("Adding title X-axis abd Y-axis label")

plt.plot(x,y)
plt.xlabel("Angle")
plt.ylabel("Sin Values")
plt.title("Sine Wave")
plt.show()
```

```python
print("parabola")

x=np.linspace(-10,10,100)
y=x**2
plt.plot(x,y)
plt.show()
```

```python
print("Color and Symbol")

plt.plot(x,y,"g.")
plt.show()
```

```python
print("Two Graphs")

x=np.linspace(-5,5,50)
plt.plot(x,np.sin(x),"g--")
plt.plot(x,np.cos(x),"r--")
plt.show()
```

```python
print("Bar Plot")

fig=plt.figure()
ax=fig.add_axes([0,0,0.8,1])
langs=["C","C++","Java","Python","PHP"]
students=[23,12,34,23,54]
ax.bar(langs,students)
plt.xlabel("Programming Language")
plt.ylabel("Number of Students")
plt.title("Stuent Survey")
plt.show()
```

```python
fig1=plt.figure()
ax1=fig1.add_axes([0,0,1,1])
langs=["C","C++","Java","Python"]
students=[12,34,56,78]
ax1.pie(students,labels=langs,autopct="%1.1f%%")
plt.show()
```

```python
print("Scatter Plot")

x=np.linspace(0,10,30)
y=np.sin(x)
z=np.cos(x)
fig2=plt.figure()
ax2=fig2.add_axes([0,0,1,1])
ax2.scatter(x,y,color="g", label="Sine")
ax2.scatter(x,z,color="r",label="Cosine")
plt.show()
```

```python
print("Scatter Plot")

fig3=plt.figure()
ax3=fig3.add_axes([0,0,1,1],projection="3d")
z=20*np.random.random(100)
x=np.sin(z)
y=np.cos(z)
ax3.scatter(x,y,z,c=z,cmap="Reds")
plt.show()
```
