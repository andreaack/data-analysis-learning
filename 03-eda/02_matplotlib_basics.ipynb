```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

# Inline display for Colab/Jupyter
%matplotlib inline
```

1. Line Chart

```python
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
sales  = [1200, 1500, 1300, 1700, 1600, 1900]

plt.figure(figsize=(8, 4))
plt.plot(months, sales, marker="o", color="steelblue", linewidth=2)
plt.title("Monthly Sales 2024")
plt.xlabel("Month")
plt.ylabel("Sales (PHP)")
plt.grid(True, linestyle="--", alpha=0.5)
plt.tight_layout()
plt.show()
```

2. Bar Chart

```python
subjects = ["Math", "English", "Science", "Filipino", "PE"]
avg_scores = [85, 78, 90, 82, 95]

plt.figure(figsize=(8, 4))
plt.bar(subjects, avg_scores, color=["steelblue", "coral", "seagreen", "gold", "mediumpurple"])
plt.title("Average Scores per Subject")
plt.xlabel("Subject")
plt.ylabel("Average Score")
plt.ylim(60, 100)
plt.tight_layout()
plt.show()
```

3. Histogram

```python
np.random.seed(42)
scores = np.random.normal(82, 10, 200)  # mean=82, std=10

plt.figure(figsize=(8, 4))
plt.hist(scores, bins=20, color="steelblue", edgecolor="white")
plt.title("Score Distribution")
plt.xlabel("Score")
plt.ylabel("Frequency")
plt.axvline(scores.mean(), color="red", linestyle="--", label=f"Mean: {scores.mean():.1f}")
plt.legend()
plt.tight_layout()
plt.show()
```

4. Scatter Plot

```python
np.random.seed(42)
math_scores    = np.random.randint(60, 100, 50)
science_scores = math_scores + np.random.randint(-15, 15, 50)

plt.figure(figsize=(6, 6))
plt.scatter(math_scores, science_scores, color="steelblue", alpha=0.7, edgecolors="white")
plt.title("Math vs Science Scores")
plt.xlabel("Math Score")
plt.ylabel("Science Score")
plt.tight_layout()
plt.show()
```

5. Multiple Subplots

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))

# Left: bar chart
axes[0].bar(subjects, avg_scores, color="steelblue")
axes[0].set_title("Average Scores")
axes[0].set_ylabel("Score")

# Right: pie chart
axes[1].pie(avg_scores, labels=subjects, autopct="%1.1f%%", startangle=90)
axes[1].set_title("Score Distribution")

plt.tight_layout()
plt.show()
```
