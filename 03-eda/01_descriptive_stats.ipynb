#  Descriptive Statistics
> **Phase 3 — EDA**  `data-analysis-practice`

Before visualizing data, you need to understand it numerically. Descriptive stats are your first look at any dataset.

---

```python
import pandas as pd
import numpy as np
```

1. Load a Dataset

```python
# We'll use a sample student performance dataset
np.random.seed(42)
df = pd.DataFrame({
    "student_id": range(1, 51),
    "math":    np.random.randint(60, 100, 50),
    "english": np.random.randint(55, 100, 50),
    "science": np.random.randint(65, 100, 50),
    "gender":  np.random.choice(["Male", "Female"], 50),
    "section": np.random.choice(["A", "B", "C"], 50)
})
df.head()
```

2. Basic Descriptive Stats

```python
# Summary statistics for all numeric columns
df.describe()
```

```python
# Individual stats
print(f"Math Mean:   {df['math'].mean():.2f}")
print(f"Math Median: {df['math'].median():.2f}")
print(f"Math Mode:   {df['math'].mode()[0]}")
print(f"Math Std:    {df['math'].std():.2f}")
print(f"Math Var:    {df['math'].var():.2f}")
print(f"Math Range:  {df['math'].max() - df['math'].min()}")
```

3. Value Counts (Categorical Data)

```python
print(df["gender"].value_counts())
print()
print(df["section"].value_counts())
```

4. Groupby : Stats by Group

```python
# Average scores by gender
print(df.groupby("gender")[["math", "english", "science"]].mean().round(2))
```

```python
# Multiple aggregations at once
summary = df.groupby("section").agg({
    "math":    ["mean", "min", "max"],
    "english": ["mean", "min", "max"],
})
print(summary)
```

5. Correlation

```python
# How related are the subjects?
correlation = df[["math", "english", "science"]].corr()
print(correlation)
```

```python
# Quick interpretation:
# Close to 1.0 = strong positive relationship
# Close to -1.0 = strong negative relationship
# Close to 0 = little to no relationship
```
