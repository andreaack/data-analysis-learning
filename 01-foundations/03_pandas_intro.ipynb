1. Import and Setup

```python
import pandas as pd
import numpy as np
print(pd.__version__)
```

2. Series  1D Data

```python
# A Series is like a labeled list
grades = pd.Series([88, 92, 78, 95, 84],
                   index=["Math", "English", "Science", "Filipino", "PE"])
print(grades)
print(grades["Math"])
print(grades[grades >= 90])
```

3. DataFrame - 2D Data (Your Main Tool)

```python
# Create a DataFrame from a dictionary
data = {
    "name":    ["Lia", "Juan", "Maria", "Pedro", "Ana"],
    "age":     [21, 22, 20, 23, 21],
    "course":  ["BSIT", "BSCS", "BSIT", "BSCE", "BSCS"],
    "gwa":     [1.75, 1.50, 2.00, 1.25, 1.75]
}

df = pd.DataFrame(data)
print(df)
```

4. Exploring a DataFrame

```python
print(df.shape)       # rows, columns
print(df.dtypes)      # data types
print(df.columns)     # column names
```

```python
df.head()    # first 5 rows
```

```python
df.tail(3)   # last 3 rows
```

```python
df.describe()  # statistical summary of numeric columns
```

```python
df.info()    # overview: nulls, dtypes, memory
```

5. Selecting Data

```python
# Select one column
print(df["name"])

# Select multiple columns
print(df[["name", "gwa"]])
```

```python
# Filter rows by condition
honor_students = df[df["gwa"] <= 1.75]
print(honor_students)
```

```python
# Multiple conditions
bsit_honor = df[(df["course"] == "BSIT") & (df["gwa"] <= 1.75)]
print(bsit_honor)
```

6. Adding and Modifying Columns

```python
# Add a new column
df["passed"] = df["gwa"] <= 3.0
df["latin_honor"] = df["gwa"].apply(lambda x: "Summa" if x <= 1.2 else
                                              "Magna" if x <= 1.5 else
                                              "Cum Laude" if x <= 1.75 else "None")
print(df)
```

7. Reading CSV Files (Most Common!)
