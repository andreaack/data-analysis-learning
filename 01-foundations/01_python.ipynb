**CONDITIONS (if / elif/ else)**

```python
grade = 88

if grade >=90:
  print("Excellent")
elif grade >=80:
  print("Good Job")
elif grade > 75:
  print("Passing")
else:
  print("Needs Improvement")
```

```python
def classify_grade(g):
  if g >= 90:
    return "A"
  elif g >= 80:
    return "B"
  elif g >= 70:
    return "C"
  elif g >= 75:
    return "D"
  else:
    return "F"

grades = [88,92,78,89,79]
for g in grades:
  print(f"{g} is {classify_grade(g)}")
```

**LOOPS**

```python
fruits = ["apple", "banana", "mango", "durian"]

for fruit in fruits:
  print(fruit.upper())


#range is used for counting
for i in range(1,6):
  print(f"Row {i}")


#while loop
count = 0
while count < 5:
  print(f"Count: {count}")
  count += 1
```

**FUNCTIONS**

```python
def greet(name):
  return f"Hello, {name}!"

print(greet("Madison"))
```

```python
#Lambda unction is a short anonymous function

double = lambda x: x*2
square = lambda x: x**2

print(double(5))
print(square(4))

numbers = [1,2,3,4,5]
doubled = list(map(lambda x: x*2, numbers))
print(doubled)
```

Practice Exercises

```python
subjects = {
    "Math":88,
    "English":72,
    "Science":90,
    "Filipino":68,
    "PE":95
}

def classify_grade(grades):
  if grades >=90:
   print("Excellent")
  elif grades >=80:
   print("Good Job")
  elif grades > 75:
   print("Passing")
  else:
   print("Needs Improvement")

for subject, score in subjects.items():
  print(f"{subject} ({score}): ", end="")
  classify_grade(score)
```
